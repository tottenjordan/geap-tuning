# Multi-run visualization (read-only)

This notebook is a **thin demo**: every step calls into the tested
`geap_tuning` package. See
[`docs/notes/doe-and-visualization.md`](../docs/notes/doe-and-visualization.md).

**No tuning cost.** It reads an experiment already populated by
[`10_doe.ipynb`](10_doe.ipynb) (or `run_experiment_tracking.py`) and charts the
runs — no jobs launched, no endpoints called. It needs live GCP **read** access
and the viz group (`uv sync --group viz`); the experiment must live in
`cfg.location`.


```python
from geap_tuning.config import load_config

EXPERIMENT_NAME = "geap-doe-sft"  # any experiment populated by an earlier run

cfg = load_config()
cfg
```

## 1. Pull the cross-run dataframe

`init_experiment` points the SDK at the region (launches nothing);
`experiment_dataframe` returns every run's params + summary metrics.
`normalize_experiment_rows` strips the `metric.`/`param.` prefixes and renames
`run_name` → `run` so the rows plot like `aggregate_results` output.

```python
from geap_tuning.experiments import experiment_dataframe, init_experiment
from geap_tuning.viz import dataframe_to_rows, normalize_experiment_rows

init_experiment(EXPERIMENT_NAME, project=cfg.project, location=cfg.location)
frame = experiment_dataframe(EXPERIMENT_NAME)
rows = normalize_experiment_rows(dataframe_to_rows(frame))
rows
```

## 2. Chart the runs

A grouped bar chart across metrics, then a single-metric bar chart.

```python
from geap_tuning.viz import plot_grouped_metric_bars

plot_grouped_metric_bars(rows)
```

```python
from geap_tuning.viz import plot_metric_bars

plot_metric_bars(rows, metric="accuracy")
```
